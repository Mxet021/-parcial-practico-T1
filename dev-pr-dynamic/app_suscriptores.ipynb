## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
# Escribe tu código aquí.
def suscriptores(n, memo=None):
    if memo is None:
        memo = {}

    if n == 0:
        return 0

    if n == 1:
        return 1

    if n in memo:
        return memo[n]

    memo[n] = suscriptores(n - 1, memo) + suscriptores(n - 2, memo)

    return memo[n]

n = int(input("Ingrese el número del mes: "))

resultado = suscriptores(n)

print("Nuevos suscriptores:", resultado)
```
